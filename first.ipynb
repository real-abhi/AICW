```python
list3 = [10,23,12,45]
list4 = []
list5 = []
for i in list3:
  if i % 2 == 0:
    print(i,"is even")
    list4.append(i)
    print(list4)
  else:
    print(i,"is odd")
    list5.append(i)
    print(list5)

#campusX
```

```python
list4 = [10,45,67,8.9,7.6,"a","b"]
list_str = []
list_int = []
list_float = []
for i in list4:
  if type(i) == int:
    list_int.append(i)
  elif type(i) == float:
    list_float.append(i)
  else:
    list_str.append(i)
print(list_int)
print(list_float)
print(list_str)
```

```python
ls = [1,2,3,4,5,6,7,8]
count_even = 0
count_odd = 0
for i in ls:
  if i % 2 ==0:
    count_even = count_even + 1
  else:
    count_odd = count_odd + 1
print(count_even)
print(count_odd)
```
